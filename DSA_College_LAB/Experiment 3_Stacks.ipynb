```python
stack=[]
while True:
    print("\n1.Push")
    print("\n2.Pop")
    print("\n3.Peek")
    print("\n4.Display")
    print("\n5.Exit")

    choice=int(input("Enter your choice:"))
    if choice==1:
        value=int(input("Enter Value:"))
        stack.append(value)
    elif choice==2:
        if len(stack)==0:
            print("Stack Underflow")
        else:
            value=stack.pop()
            print(value)
    elif choice==3:
        if len(stack)==0:
            print("Stack underflow")
        else:
            print("The top element is ", stack[-1])         
    elif choice==4:
        if len(stack)==0:
            print("Stack is Empty")
        else:
            print("Stack Elements:")
            for item in stack:
                print(item)
    elif choice==5:
        break
    else:
        print("Invalid Input!")                                   
```

```python
word=input()
stack=[]
for ch in word:
    stack.append(ch)
reverse=""
while len(stack)>0:
    reverse=reverse+stack.pop()
print(reverse)        
```
